# QUESTION 2: Text Cleaning and Processing

* Solution of 2(a).1: Loading the CSV File and finding review count

```python
import pandas as pd
data = pd.read_csv('IMDB Dataset.csv')
print(data.head())
```

* Solution of 2(a).2: Count the Total Number of Reviews

```python
total_reviews = len(data)
print("Total number of reviews:", total_reviews)
```

* Find the Longest Review (Words):

```python
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')

longest_words = max(data['review'], key=lambda x: len(word_tokenize(x)))
longest_words_index = data['review'].tolist().index(longest_words)

print("\nLongest review by words:")
print("Index:", longest_words_index)
print("Word count:", len(word_tokenize(longest_words)))
print("Review:")
print(longest_words)
```

* Step 4: Find the Longest Review (Sentences)

```python
from nltk.tokenize import sent_tokenize

longest_sentences = max(data['review'], key=lambda x: len(sent_tokenize(x)))
longest_sentences_index = data['review'].tolist().index(longest_sentences)

print("\nLongest review by sentences:")
print("Index:", longest_sentences_index)
print("Sentence count:", len(sent_tokenize(longest_sentences)))
print("Review:")
print(longest_sentences)
```

* 2(b): Cleaning Step 1 - Remove Stopwords and Punctuation

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

cleaned_reviews = []
for review in data['review']:
    tokens = word_tokenize(review)
    tokens = [word.lower() for word in tokens if word.isalnum() and word.lower() not in stop_words]
    cleaned_reviews.append(" ".join(tokens))
```

* Proof 1 that 2b worked fine: Original Review vs Cleaned Review (Pick the first review as an example)

```python
print("Original Review (First):", data['review'][0])
print("Cleaned Review (First):", cleaned_reviews[0])
```

* Proof 2 of 2b code working fine: Count of Reviews Before and After Cleaning (to verify no reviews were lost).

```python
print("Total reviews before cleaning:", len(data))
print("Total reviews after cleaning:", len(cleaned_reviews))
```

* 2(c): Cleaning Step 2 - Save to File

```python
with open('/content/cleaned_reviews.txt', 'w') as f:
    for review in cleaned_reviews:
        f.write(review + "\n")
```

# QUESTION 3: Representation

* 3(a) TF-IDF Vectorization

* 3a.1: Load Cleaned Dataset:

```python
with open('cleaned_reviews.txt', 'r') as f:
    cleaned_reviews = f.readlines()

cleaned_reviews = [review.strip() for review in cleaned_reviews]

print("Number of cleaned reviews:", len(cleaned_reviews))
print("Sample review:", cleaned_reviews[0])
```

*  3a.2: Perform TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(cleaned_reviews)

feature_names = vectorizer.get_feature_names_out()

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Sample features (words):", feature_names[:10])
```

* 3a.3 Display Results

```python
tfidf_df = pd.DataFrame.sparse.from_spmatrix(tfidf_matrix, columns= feature_names)

print(tfidf_df.head())
print(tfidf_df.shape)
```

```python
print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Sample features (words):", feature_names[:10])
first_review_vector = tfidf_matrix[0].toarray()[0]

non_zero = [(feature_names[i], first_review_vector[i]) for i in range(len(feature_names)) if first_review_vector[i] > 0]
print("Non-zero TF-IDF values (first review):", non_zero[:10])
```

* Solution of 3(b): Word Embeddings and Cosine Similarities



* Solution of 3b.1: Download and load GLoVe embeddings:

```python
import numpy as np
embeddings = {}
with open('glove.6B.100d.txt', 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.asarray(values[1:], dtype='float32')
        embeddings[word] = vector
```

* Solution of 3b.2: Compute Cosine Similarity:

```python
from numpy.linalg import norm

def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))

print("Man-Woman:", cosine_similarity(embeddings['man'], embeddings['woman']))
print("Cat-Dog:", cosine_similarity(embeddings['cat'], embeddings['dog']))
print("King-Queen:", cosine_similarity(embeddings['king'], embeddings['queen']))
```

* Solution of 3b.3: Verify the Equation:

```python
result_vector = embeddings['king'] - embeddings['man'] + embeddings['woman']
similarity = cosine_similarity(result_vector, embeddings['queen'])
print("Similarity (King - Man + Woman vs Queen):", similarity)
```
